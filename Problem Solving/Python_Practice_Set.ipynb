## Question 1 — Expressions, Variables & Conditions

A store wants to calculate a shipping fee based on the order weight (in kg).

### Rules
- If weight is **more than 20 kg**, fee is `weight * 15`.
- If weight is **between 5 and 20 kg (inclusive)**, fee is `weight * 10`.
- Otherwise (less than 5 kg), fee is a **flat 25**.
- Return the fee rounded to 2 decimal places.

### Example

**Input**
```text
weight = 12
```

**Expected Output**
```text
120.0
```


````python

def calculate_shipping_fee(weight):
    if weight >= 20:
        return weight*15
    elif weight >= 5:
        return weight * 10
    else:
        return 25
    


# Test input
weight = 12
result = calculate_shipping_fee(weight)

print(result)
````


## Question 2 — Loops, Functions & Libraries

Write the logic to find all numbers between 1 and n (inclusive) that are **prime**,
and return them as a list. Use a loop-based approach — no need for optimized algorithms.

### Example

**Input**
```text
n = 20
```

**Expected Output**
```text
[2, 3, 5, 7, 11, 13, 17, 19]
```


````python
y = []

def find_primes(n):
    for i in range(2,n):
        for z in range(2,i):
           if i % z == 0:
            break
        else:
            y.append(i)
    return y
           

# Test input
n = 20
result = find_primes(n)

print(result)
````

Complete the method `capacity_status()`:

- If usage is **90 or more**, return `"Critical"`.
- If usage is **60 or more but less than 90**, return `"Busy"`.
- Otherwise, return `"Available"`.

### Example

**Input**
```text
Name: Warehouse-A
Usage: 75
```

**Expected Output**
```text
Busy
```


````python

class Warehouse:
    def __init__(self, name, usage):
        self.name = name
        self.usage = usage

    def capacity_status(self):
        # YOUR LOGIC HERE
        if self.usage >=90:
            return "Critical"
        elif self.usage >=60 and self.usage < 90:
            return "Busy"
        else:
            return "Available"
        pass


# Test input
wh = Warehouse("Warehouse-A", 75)

print(wh.capacity_status())
````
