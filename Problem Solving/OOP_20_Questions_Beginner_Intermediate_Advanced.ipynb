## Q10 – ATM System

Use inheritance, encapsulation and overriding. HDFC limit=20000, AXIS limit=30000. Withdrawal requires enough balance and must be within the bank limit.

```python
class Bank:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        # YOUR LOGIC HERE
        return self.__balance
        pass

class HDFC(Bank):
    def withdraw(self, amount):
        # YOUR LOGIC HERE
        balance=self.get_balance()
        if amount <= balance:
            w=balance-amount
            print("HDFC withdrawal successful")
            print("Avilable Balance :%d"%(w))
        else:
            print("insufficient Balance!")    
            
        
        pass

class AXIS(Bank):
    def withdraw(self, amount):
        # YOUR LOGIC HERE
        balance=self.get_balance()
        if amount <= balance:
            w=balance-amount
            print("AXIS withdrawal successful")
            print("Avilable Balance :%d"%(w))
        else:
            print("insufficient Balance!")  
    
        
        pass

banks = [HDFC(50000), AXIS(50000)]
amounts = [15000, 25000]
for bank, amount in zip(banks, amounts):
    bank.withdraw(amount)

# INPUT: HDFC withdrawal=15000, AXIS withdrawal=25000
# EXPECTED OUTPUT:
# HDFC withdrawal successful
# AXIS withdrawal successful
```

## Q12 – Shape Abstraction

Create abstract `Shape`. `Circle.area()` uses 3.14*r*r; `Rectangle.area()` uses length*width. Use one loop.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pi=3.14
        return pi
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        # YOUR LOGIC HERE
       return super().area()*self.radius*self.radius
    pass

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        # YOUR LOGIC HERE
        return self.length*self.width
       
        pass

shapes = [Circle(7), Rectangle(10, 5)]
for shape in shapes:
    print(shape.area())

# INPUT: Circle radius=7; Rectangle=10x5; pi=3.14
# EXPECTED OUTPUT:
# 153.86
# 50
```

## Q14 – Shopping Cart

Use composition. `ShoppingCart` stores Product objects. Complete `add_product()` and `total_price()`.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class ShoppingCart:
    def __init__(self):
        self.products = []
        

    def add_product(self, product):
        # YOUR LOGIC HERE
        self.products.append(product)
        pass

    def total_price(self):
        # YOUR LOGIC HERE
        total=0
        for product in self.products:
            total += product.price
        return total    
        pass

cart = ShoppingCart()
cart.add_product(Product("Keyboard", 1500))
cart.add_product(Product("Mouse", 500))
cart.add_product(Product("USB", 300))
print("Total:", cart.total_price())

# INPUT: 1500, 500, 300
# EXPECTED OUTPUT:
# Total: 2300
```

## Q18 – E-Commerce Product

Combine abstraction, inheritance, encapsulation and polymorphism. Keep stock private. Electronics tax=18%, Clothing tax=5%. Complete buy, get_stock and final_price.

```python
from abc import ABC, abstractmethod

class Product(ABC):
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.__stock = stock

    def buy(self, quantity):
        # YOUR LOGIC HERE
        self.__stock -= quantity
        print("Purchase successful")
        
        pass

    def get_stock(self):
        # YOUR LOGIC HERE
        return self.__stock
        pass

    @abstractmethod
    def final_price(self):
        pass

class Electronics(Product):
    def final_price(self):
        # YOUR LOGIC HERE
        e=self.price*0.18
        self.price +=e
        return self.price
        pass

class Clothing(Product):
    def final_price(self):
        # YOUR LOGIC HERE
        c=self.price*0.05
        self.price +=c
        return self.price
        pass

products = [Electronics("Laptop", 50000, 5), Clothing("Jacket", 2000, 10)]
for product in products:
    print(product.name, product.final_price())

products[0].buy(2)
print("Laptop stock:", products[0].get_stock())

# INPUT: Laptop=50000 stock=5; Jacket=2000 stock=10; buy 2 laptops
# EXPECTED OUTPUT:
# Laptop 59000.0
# Jacket 2100.0
# Purchase successful
# Laptop stock: 3
```

## Q19 – Ride Booking

Use an abstract Ride class. Bike=₹10/km, Car=₹15/km + ₹50 base, Auto=₹12/km. Calculate each fare and the total.

```python
from abc import ABC, abstractmethod

class Ride(ABC):
    def __init__(self, distance):
        self.distance = distance
    @abstractmethod
    def calculate_fare(self):
        pass

class Bike(Ride):
    def calculate_fare(self):
        # YOUR LOGIC HERE
        
        p=self.distance*10
        return p
        pass

class Car(Ride):
    def calculate_fare(self):
        # YOUR LOGIC HERE
        t=50
        t=t+self.distance*15
        return t
        pass

class Auto(Ride):
    def calculate_fare(self):
        # YOUR LOGIC HERE
        a=0
        a=self.distance*12
        return a
        pass

rides = [Bike(10), Car(10), Auto(10)]
total = 0
for ride in rides:
    fare = ride.calculate_fare()
    print("Fare:", fare)
    total += fare
print("Total Fare:", total)

# INPUT: all rides = 10 km
# EXPECTED OUTPUT:
# Fare: 100
# Fare: 200
# Fare: 120
# Total Fare: 420
```

## Q20 – Online Banking System

Combine abstraction, encapsulation, inheritance and polymorphism. Savings interest=5%, Current interest=2%. Complete deposit, withdraw, get_balance and calculate_interest.

```python
from abc import ABC, abstractmethod

class BankAccount(ABC):
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def deposit(self, amount):
        # YOUR LOGIC HERE
        self.__balance=self.__balance + amount
        pass

    def withdraw(self, amount):
        # YOUR LOGIC HERE
        self.__balance=self.__balance-amount
        pass

    def get_balance(self):
        # YOUR LOGIC HERE
        return self.__balance
        pass

    @abstractmethod
    def calculate_interest(self):
        pass

class SavingsAccount(BankAccount):
    def calculate_interest(self):
        # YOUR LOGIC HERE
        return self.get_balance()*0.05
        pass

class CurrentAccount(BankAccount):
    def calculate_interest(self):
        # YOUR LOGIC HERE
        return self.get_balance()*0.02
        
        pass

accounts = [SavingsAccount("Yash", 100000), CurrentAccount("Rahul", 100000)]
for account in accounts:
    print(account.name)
    print("Interest:", account.calculate_interest())

# INPUT: Yash Savings=100000; Rahul Current=100000
# EXPECTED OUTPUT:
# Yash
# Interest: 5000.0
# Rahul
# Interest: 2000.0
```
